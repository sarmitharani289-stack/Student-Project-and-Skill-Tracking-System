## 1. Environment Setup and Database Retrieval
Connect to the SQLite database `student_project_tracking.db` and query entities into data structures.

```python
import sqlite3
import os

db_path = os.path.join("..", "database_management", "student_project_tracking.db")
conn = sqlite3.connect(db_path)
cursor = conn.cursor()
print("[+] Connected successfully to:", db_path)
```

## 2. Data Preprocessing & Quality Verification
Verify null constraints and sanitize records.

```python
cursor.execute("SELECT COUNT(*) FROM Student;")
print("Total Students:", cursor.fetchone()[0])

cursor.execute("SELECT COUNT(*) FROM Project;")
print("Total Registered Projects:", cursor.fetchone()[0])

cursor.execute("SELECT COUNT(*) FROM Milestone;")
print("Total Milestones Tracked:", cursor.fetchone()[0])

cursor.execute("SELECT COUNT(*) FROM Evaluation;")
print("Evaluated Projects:", cursor.fetchone()[0])
```

## 3. Pattern 1: Project Completion Rates
Analyze overall project progression status across proposed, in-progress, and completed states.

```python
cursor.execute("""
SELECT status, COUNT(*) as count, ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM Project), 1) as pct
FROM Project
GROUP BY status;
""")
for row in cursor.fetchall():
    print(f"{row[0]}: {row[1]} projects ({row[2]}%)")
```

## 4. Pattern 2: Popular Technology Choices
Frequency analysis of languages, frameworks, and tools selected by students for mini-projects.

```python
cursor.execute("""
SELECT technology_name, COUNT(*) as count
FROM ProjectTechnology
GROUP BY technology_name
ORDER BY count DESC;
""")
for row in cursor.fetchall():
    print(f"{row[0]:<15}: {row[1]} projects")
```

## 5. Pattern 3: Skill Distribution and Matrix
Breakdown of student competencies across technical categories and proficiency levels.

```python
cursor.execute("""
SELECT sk.skill_category, 
       SUM(CASE WHEN ss.proficiency_level = 'Advanced' THEN 1 ELSE 0 END) AS adv,
       SUM(CASE WHEN ss.proficiency_level = 'Intermediate' THEN 1 ELSE 0 END) AS inter,
       SUM(CASE WHEN ss.proficiency_level = 'Beginner' THEN 1 ELSE 0 END) AS beg
FROM Skill sk
LEFT JOIN StudentSkill ss ON sk.skill_id = ss.skill_id
GROUP BY sk.skill_category;
""")
for row in cursor.fetchall():
    print(f"{row[0]:<16} | Advanced: {row[1]} | Intermediate: {row[2]} | Beginner: {row[3]}")
```
